```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')
```

```python
sales = pd.read_csv("sales_data.csv")
customers = pd.read_csv("customer_churn.csv")
```

```python
sales.head()
customers.head()
```

Step 6: Explore Dataset

```python
print(sales.shape)
print(customers.shape)

print(sales.info())
print(customers.info())
```

Missing values:

```python
print(sales.isnull().sum())
print(customers.isnull().sum())
```

Step 7: Data Cleaning

```python
sales['Date'] = pd.to_datetime(sales['Date'])
```

```python
sales['Year'] = sales['Date'].dt.year
sales['Month'] = sales['Date'].dt.month
sales['Day'] = sales['Date'].dt.day
```

Step 8: Customer Analysis

```python
top_customers = sales.groupby('Customer_ID')['Total_Sales'].sum()

top_customers = top_customers.sort_values(
    ascending=False
)

print(top_customers.head(10))
```

Step 9: Monthly Sales Analysis

```python
monthly_sales = sales.groupby('Month')['Total_Sales'].sum()

print(monthly_sales)
```

Visualization:

```python
monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(8,5)
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

Step 10: Product Analysis

```python
product_sales = sales.groupby('Product')['Quantity'].sum()

product_sales = product_sales.sort_values(
    ascending=False
)

print(product_sales)
```

Visualization:

```python
product_sales.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title("Best Selling Products")
plt.show()
```

Step 11: Regional Analysis

```python
region_sales = sales.groupby('Region')['Total_Sales'].sum()

print(region_sales)
```

Visualization:

```python
region_sales.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title("Sales By Region")
plt.ylabel("")
plt.show()
```

Step 12: Multiple Condition Filtering

```python
filtered_data = sales[
    (sales['Region']=='North') &
    (sales['Total_Sales']>5000)
]

filtered_data.head()
```

Step 13: String Operations

```python
sales['Product_Upper'] = sales['Product'].str.upper()

sales['Product_Length'] = sales['Product'].str.len()

sales.head()
```

Step 14: Customer Churn Analysis

```python
churn_rate = customers['Churn'].mean()*100

print("Churn Rate:", churn_rate)
```

Contract-wise churn:

```python
customers.groupby('Contract')['Churn'].mean()
```

Step 15: Merge Dataset

```python
customers.rename(
    columns={'CustomerID':'Customer_ID'},
    inplace=True
)
```

```python
merged = pd.merge(
    sales,
    customers,
    on='Customer_ID',
    how='left'
)

merged.head()
```

Step 16: Create Pivot Table

```python
pivot_table = pd.pivot_table(
    sales,
    values='Total_Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0
)

print(pivot_table)
```

Step 17: Key Business Metrics

```python
total_revenue = sales['Total_Sales'].sum()

total_customers = sales['Customer_ID'].nunique()

average_order_value = sales['Total_Sales'].mean()

print("Revenue:", total_revenue)
print("Customers:", total_customers)
print("AOV:", average_order_value)
```
